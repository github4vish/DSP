```python
# Step 1: Import Libraries
import pandas as pd
from surprise import Dataset, Reader, SVD
from surprise.model_selection import train_test_split
from surprise import accuracy

# Step 2: Load Dataset
ratings = pd.read_csv("ratings.csv")

# Step 3: Prepare Data for Surprise Library
reader = Reader(rating_scale=(1, 5))
data = Dataset.load_from_df(ratings[['UserID', 'ItemID', 'Rating']], reader)

# Step 4: Train-Test Split
trainset, testset = train_test_split(data, test_size=0.2, random_state=42)

# Step 5: Train Model (SVD)
model = SVD()
model.fit(trainset)

# Step 6: Predictions
predictions = model.test(testset)

# Step 7: Evaluation
print("RMSE:", accuracy.rmse(predictions))
print("MAE:", accuracy.mae(predictions))

# Step 8: Recommend Top Items for a User
user_id = 1
items = ratings['ItemID'].unique()

print("\nTop Recommendations for User 1:")
for item in items:
    pred = model.predict(user_id, item)
    print("Item:", item, "Predicted Rating:", round(pred.est, 2))
```
